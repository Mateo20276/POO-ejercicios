**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

# ***The answers about customers did not refer to unique customers. However, they were printed in questions that made sense to print.***

```python
from os.path import exists
import csv

# check if file can be found and open the file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

# customers file is in the sample_data folder and can be opened.
# Using csv reader is one way of dealing with the file
# but it's up to you how you want to proceed

customer_ids = set()

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header

    total_customers = 0

    for row in csvreader:
        total_customers += 1
        customer_ids.add(row[0])

print("There are", total_customers, "customers")
print("There are", len(customer_ids), "unique customers")

print("All done!")
```

**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
# evaluate the contents of the file

different_states = []

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header

    for row in csvreader:
        state = row[4].strip().upper()

        if state not in different_states:
            different_states.append(state)

#print(different_states)
print("Different states:", len(different_states))
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
state_customers = {}
states_unique_customers = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header

    for row in csvreader:
        customer_id = row[0]
        state = row[4].strip().upper()

        if state in state_customers:
            state_customers[state] += 1
            states_unique_customers[state].add(customer_id)
        else:
            state_customers[state] = 1
            states_unique_customers[state] = set([customer_id])

max_state = max(state_customers, key=state_customers.get)

max_unique_state = max(
    states_unique_customers,
    key=lambda x: len(states_unique_customers[x])
)

print("State with the most customers:", max_state)
print("Number of customers:", state_customers[max_state])

print("State with the most unique customers:", max_unique_state)
print("Number of unique customers:", len(states_unique_customers[max_unique_state]))
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file
state_customers = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header

    for row in csvreader:
        state = row[4].strip().upper()

        if state in state_customers:
            state_customers[state] += 1
        else:
            state_customers[state] = 1

min_customers = min(state_customers.values())

states_with_min = [state for state, count in state_customers.items() if count == min_customers]

print("States with the least customers")
for state in states_with_min:
    print(state)

print("Number of customers", min_customers)
```

**Question 5:** What is the most common last name?

```python
# open file and init vars
# evaluate the contents of the file
lastname_customers = {}
unique_lastname_customers = {}

with open(datafile, 'r') as fl:
    csvreader = csv.reader(fl, delimiter=',')
    next(csvreader)  # Skip header

    for row in csvreader:
        customer_id = row[0]
        lastname = row[2].strip().upper()

        if lastname in lastname_customers:
            lastname_customers[lastname] += 1
            unique_lastname_customers[lastname].add(customer_id)
        else:
            lastname_customers[lastname] = 1
            unique_lastname_customers[lastname] = set([customer_id])

#print(lastname_customers)
max_lastname = max(lastname_customers, key=lastname_customers.get)
max_unique_lastname = max(
    unique_lastname_customers,
    key=lambda x: len(unique_lastname_customers[x])
)

print("Last name with the most customers:", max_lastname)
print("Number of last name:", lastname_customers[max_lastname])

print("Last name with the most unique customers:", max_unique_lastname)
print("Number of unique last name:", len(unique_lastname_customers[max_unique_lastname]))
```

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

```python
# open and evaluate orders.csv
from os.path import exists
import csv

# check if file can be found and open the file
datafile = "./sample_data/orders.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

unique_orders = set()

with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  next(csvreader)  # Skip header

  orders = list(csvreader)
  unique_orders = set([order[1] for order in orders])

print("Question #1")
print("There are", len(unique_orders), "unique orders")
print("\n")

items_per_order = {}

for order in orders:
    order_id = order[1]

    if order_id in items_per_order:
        items_per_order[order_id] += 1
    else:
        items_per_order[order_id] = 1

average_items = sum(items_per_order.values()) / len(items_per_order)

print("Question #2")
print("The average number of items per order is", round(average_items, 2))
print("\n")

highest_items = max(items_per_order.values())

print("Question #3")
print("The highest number of items per order is", highest_items)
print("\n")

OCTOBER_2021 = "2021-10"
october_orders = 0

for order in orders:
    date = order[2]

    if date.startswith(OCTOBER_2021):
        october_orders += 1

print("Question #4")
print("The number of orders placed in October 2021 are", october_orders)
print("\n")

customer_spent = {}
seen_orders = set()

for order in orders:
    date = order[2]

    if not date.startswith("2021"):
        continue

    order_id = order[1]

    if order_id in seen_orders:
        continue

    seen_orders.add(order_id)

    customer_id = order[0]
    total = float(order[3])

    if customer_id in customer_spent:
        customer_spent[customer_id] += total
    else:
        customer_spent[customer_id] = total

max_customer = max(customer_spent, key=customer_spent.get)

print("Question #5")
print("The customer spent the most amount of money in 2021 is", max_customer)
print("Total spent", customer_spent[max_customer])
print("\n")

month_sales = {}
seen_orders = set()

for order in orders:
    order_id = order[1]
    date = order[2]

    if order_id in seen_orders or not date:
        continue

    seen_orders.add(order_id)

    month = date[:7].split("-")[1]
    total = float(order[3])

    if month in month_sales:
        month_sales[month] += total
    else:
        month_sales[month] = total

best_month = max(month_sales, key=month_sales.get)

print("Question #6")
print("Historically, the best month for sales is", best_month)
print("Total spent", month_sales[best_month])
print("\n")


```
